```python
import json
import re

# Load the notebook file
notebook_path = '/content/1.1.3Indexing_and_Slicing_in_Python_.ipynb'
with open(notebook_path, 'r', encoding='utf-8') as f:
    nb_data = json.load(f)

def remove_comments(code_str):
    lines = code_str.split('\n')
    clean_lines = []
    for line in lines:
        # Simple comment removal: strip line-level and trailing comments
        # Handles basic Python comment lines starting with #
        stripped = line.strip()
        if stripped.startswith('#'):
            continue
        # Remove trailing comments where # is preceded by whitespace
        # to avoid stripping inside strings (as a simple heuristic)
        line_no_comment = re.sub(r'\s+#.*$', '', line)
        clean_lines.append(line_no_comment)
    return '\n'.join(clean_lines)

# Extract and clean code from code cells
for idx, cell in enumerate(nb_data.get('cells', [])):
    if cell.get('cell_type') == 'code':
        source = cell.get('source', [])
        if isinstance(source, list):
            code_content = ''.join(source)
        else:
            code_content = source

        clean_code = remove_comments(code_content).strip()
        if clean_code:
            print(f"--- Executing Cell {idx+1} ---")
            exec(clean_code)
```

```python
import json
import re

with open('/content/1.1.3Indexing_and_Slicing_in_Python_.ipynb', 'r', encoding='utf-8') as f:
    nb_data = json.load(f)

def clean_cell_code(code_str):
    lines = code_str.split('\n')
    clean_lines = []
    for line in lines:
        stripped = line.strip()
        if stripped.startswith('#'):
            continue
        line_no_comment = re.sub(r'\s+#.*$', '', line)
        clean_lines.append(line_no_comment)
    return '\n'.join(clean_lines).strip()

# Filter out only valid, non-empty code cells
code_blocks = []
for cell in nb_data.get('cells', []):
    if cell.get('cell_type') == 'code':
        source = cell.get('source', [])
        code_content = ''.join(source) if isinstance(source, list) else source
        clean_code = clean_cell_code(code_content)
        if clean_code:
            code_blocks.append(clean_code)

# We will dynamically execute the extracted code blocks sequentially in this session
for idx, block in enumerate(code_blocks):
    print(f"\n--- Executing Extracted Program Cell {idx+1} ---")
    exec(block)
```

### **Part 1: Lists**

```python
my_list = [10, 20, 30, 40, 50]
```

```python
print("First element:", my_list[0])
print("Last element:", my_list[-1])
```

```python
print("Elements from index 1 to 3 (not inclusive):", my_list[1:4])
print("From start to index 2:", my_list[:3])
print("From index 2 to end:", my_list[2:])
print("Every second element:", my_list[::2])
print("Reversed list:", my_list[::-1])
```

### **Part 2: Tuples**

```python
my_tuple = ("a", "b", "c", "d", "e")
```

```python
print("Element at index 0:", my_tuple[0])
print("Element at index -2:", my_tuple[-2])
```

```python
print("Slice from index 1 to 3:", my_tuple[1:4])
print("Tuple reversed:", my_tuple[::-1])
```
